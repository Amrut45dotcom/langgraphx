```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class Batsmanstate(TypedDict):
    balls:int
    runs:int
    fours:int
    sixes:int
    sr:float
    bpb:float
    boundary_percentage:float
    summary:str
```

```python
def calculate_sr(state:Batsmanstate)->Batsmanstate:
    state['sr'] = (state['runs']/state['balls'])*100
    return {'sr': state['sr']}
```

```python
def calculate_bpb(state:Batsmanstate)->Batsmanstate:
    state['bpb'] = state['balls']/(state['fours']+state['sixes'])
    return {'bpb': state['bpb']}
```

```python
def calculate_boundary_percentage(state:Batsmanstate)->Batsmanstate:
    state['boundary_percentage'] = ((state['fours']*4 + state['sixes']*6)/state['runs'])*100
    return {'boundary_percentage': state['boundary_percentage']}
    
```

```python
def summarize(state:Batsmanstate)->Batsmanstate:
    state['summary'] = f"Runs: {state['runs']}, Balls: {state['balls']}, Fours: {state['fours']}, Sixes: {state['sixes']}, SR: {state['sr']:.2f}, BPB: {state['bpb']:.2f}, Boundary Percentage: {state['boundary_percentage']:.2f}%"
    return state
```

```python
graph = StateGraph(Batsmanstate)
graph.add_node('calculate_sr',calculate_sr)
graph.add_node('calculate_bpb',calculate_bpb)
graph.add_node('calculate_boundary_percentage',calculate_boundary_percentage) 
graph.add_node('summarize',summarize)


#edges
graph.add_edge(START,'calculate_sr')
graph.add_edge(START,'calculate_bpb')
graph.add_edge(START,'calculate_boundary_percentage')

graph.add_edge('calculate_sr','summarize')
graph.add_edge('calculate_bpb','summarize') 
graph.add_edge('calculate_boundary_percentage','summarize')

graph.add_edge('summarize',END)

workflow = graph.compile()
```

```python
initial_state = {
    'balls': 29,
    'runs': 97,
    'fours': 8,
    'sixes': 12
}
final_state = workflow.invoke(initial_state)
print(final_state['summary'])   
```
