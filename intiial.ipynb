```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
#define state for BMI calculation
class BMIstate(TypedDict):
    weight: float
    height: float
    bmi: float
    obese: bool
```

```python
def calculate_bmi(state: BMIstate) -> BMIstate:
    bmi = state['weight'] / (state['height'] ** 2)
    state['bmi'] = round(bmi, 2)
    return state
```

```python
def check_obesity(state: BMIstate) -> BMIstate:
    state['obese'] = state['bmi'] >= 30
    return state
```

```python
graph = StateGraph(BMIstate)
graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('check_obesity', check_obesity)
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'check_obesity')
graph.add_edge('check_obesity', END)

workflow = graph.compile()
```

```python
print(workflow.invoke({
    'weight': 70,
    'height': 1.75
}))
```
